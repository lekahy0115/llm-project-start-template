# Drive 연결 및 경로 확인
처음 실행할 때 Google Drive를 연결하고 경로만 확인합니다. Git 명령을 실행하거나 파일을 수정하지 않습니다.


```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
from pathlib import Path

# 팀 정보에 맞게 다음 두 이름을 수정합니다.
PROJECT_FOLDER = 'RENAME_ME_PROJECT'
SHARED_DATA_FOLDER = 'RENAME_ME_SHARED_DATA'

MY_DRIVE = Path('/content/drive/MyDrive')
REPO = MY_DRIVE / PROJECT_FOLDER / 'repository'
SHARED_DATA = MY_DRIVE / SHARED_DATA_FOLDER

print('개인 Git 저장소:', REPO, '존재:', (REPO / '.git').is_dir())
print('공용 대용량 폴더:', SHARED_DATA, '존재:', SHARED_DATA.is_dir())
print('노트북 폴더:', REPO / 'notebooks', '존재:', (REPO / 'notebooks').is_dir())
```
