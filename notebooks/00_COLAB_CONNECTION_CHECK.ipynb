# Colab / Google Drive 연결 확인

이 노트북은 팀 공통 환경을 확인하기 위한 **연결 점검용** 노트북입니다. LLM 기능이나 서비스 코드는 포함하지 않습니다.

1. 아래 첫 셀을 실행해 Google Drive를 마운트합니다.
2. 둘째 셀의 `DRIVE_FOLDER_NAME`을 **팀 공통으로 사용하는 실제 공유 폴더명**으로 변경합니다.
3. 둘째 셀을 실행하여 폴더 존재 여부를 확인합니다.

팀원용 설명: `docs/TEAM_COMMON_SETUP.md`

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
from pathlib import Path

# 팀의 실제 Google Drive 공유 폴더 이름으로 변경하세요.
DRIVE_FOLDER_NAME = 'RENAME_ME_SHARED_DRIVE_FOLDER'
DRIVE_ROOT = Path('/content/drive/MyDrive') / DRIVE_FOLDER_NAME

if DRIVE_FOLDER_NAME.startswith('RENAME_ME_'):
    raise ValueError('DRIVE_FOLDER_NAME을 실제 공유 폴더 이름으로 변경하세요.')

print('공유 폴더:', DRIVE_ROOT)
print('폴더 존재:', DRIVE_ROOT.is_dir())

for name in ['datasets/raw', 'datasets/processed', 'models', 'checkpoints', 'outputs']:
    print(f'{name}:', (DRIVE_ROOT / name).is_dir())

if not DRIVE_ROOT.is_dir():
    raise FileNotFoundError('공유 폴더가 없습니다. Drive 접근 권한/내 드라이브 바로가기/폴더 이름을 확인하세요.')
```
